```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("pinky")
greet_user("sweety")
```

```python
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")
greet_user()
greet_user("Ravi")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("pinky", "sweety", "tinky")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=20, city="gujarat")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="rakesh", city="Mumbai")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 9:", square(9))
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]
print(list(map(lambda x: x ** 2, numbers)))
print(list(filter(lambda x: x % 2 == 0, numbers)))
print(reduce(lambda x, y: x + y, numbers))
```

```python
student = {
    "bob": 70,
    "alice": 80,
    "John": 75
}

result = list(map(lambda x: x.upper(), student))

print(result)
```
